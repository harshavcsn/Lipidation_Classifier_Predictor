## 2. Enter your sequences

Each sequence must be 4 residues (anchor `G` + positions A2, A3, A4) using the 20 standard amino acids. Separate sequences with commas, spaces or new lines.

```python
sequences_text = "GAAC, GKLV, GGWV"  #@param {type:"string"}

import re
sequences = [s for s in re.split(r"[\s,;]+", sequences_text) if s]
results = predict(sequences)
results
```

## 3. (Optional) Upload a file of sequences

One sequence per line with no header, or a CSV with a `sequence` column. Skip this cell if you entered sequences above.

```python
from google.colab import files

uploaded = files.upload()
filename = next(iter(uploaded))
results = predict(read_sequence_file(filename))
results
```

## 4. Download the results

```python
results.to_csv("predictions.csv", index=False)
files.download("predictions.csv")
```

## Reading the output

| column | meaning |
|---|---|
| `<model>_pred` | label predicted by that model |
| `consensus_pred` | majority vote across the models |
| `consensus_agreement` | how many models agree with the consensus; treat split votes such as `2/3` as low-confidence |

To use a single model, call `predict(sequences, models=["SVM_RBF"])`.
