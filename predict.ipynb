# Lipidation Classifier: predict your sequences

Classifies G-X-X-X tetrapeptides as **fiber**, **droplet** or **MS** using the three trained models in `results_4block_pca/final/` (SVM, elastic-net logistic regression, random forest), plus a majority-vote consensus.

Run this notebook from the repository root (the folder containing `predict.py`). Install the dependencies first:

```bash
pip install -r requirements.txt
```

````python
from predict import predict, read_sequence_file
````

## 1. Predict a few sequences

Each sequence must be 4 residues (anchor `G` + positions A2, A3, A4) using the 20 standard amino acids. Lower case is fine.

````python
sequences = ["GAAC", "GKLV", "GGWV"]

results = predict(sequences)
results
````

## 2. Predict sequences from a file

Either one sequence per line with no header (like `new_predict.csv`), or a CSV with a `sequence` column.

````python
results = predict(read_sequence_file("new_predict.csv"))
results.head(10)
````

## 3. Save the results

````python
results.to_csv("my_predictions.csv", index=False)
print(f"Saved {len(results)} rows to my_predictions.csv")
````

## Options

* **Use one model only:** `predict(sequences, models=["SVM_RBF"])`. SVM_RBF had the best cross-validated macro-F1.
* **Use your own retrained models:** `predict(sequences, models_dir="my_run/final", pca_table_path="my_run/features/pca_table.csv")`

**Reading the output:** `consensus_agreement` shows how many models agree with the consensus. Treat split votes (e.g. `2/3`) as low-confidence calls.
